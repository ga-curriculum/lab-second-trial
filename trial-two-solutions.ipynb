## Problem 1: Odd Numbers

**Step 1:** Given the following list of integers, create a new list that is this list filtered down to only odd numbers. **This should be done in one line of code with a list comprehension.**

````python
my_numbers = [
    8, 71, 70, 95, 89, 25, 16, 62, 93, 80,
    24, 56, 32, 72, 29, 100, 23, 100, 58, 74,
    19, 37, 18, 14, 26, 89, 13, 84, 7, 86
]
````

````python
odd_numbers = [n for n in my_numbers if n % 2 == 1]
````

**Step 2:** Calculate the sum of the odd numbers in the above list. (Yes this is as easy as it sounds, feel free to use a built-in function.)

````python
sum(odd_numbers)
````

**Step 3:** Create a function called `odd_sum(nums)` that takes in a list of numbers (integers) and returns the sum of all odd numbers in that list. Test out your function on the list provided in Step 1.

````python
def odd_sum(nums):
    odd_nums = [n for n in nums if n % 2 == 1]
    return sum(odd_nums)

odd_sum(my_numbers)
````

**Step 2:** Write a function `sum_sq()` that takes in an integer and returns the sum of the squares of all numbers up until that number.

```
sum_sq(5) = 1 + 4 + 9 + 16 + 25 = 55
```

````python
def sum_sq(n):
    return sum([i**2 for i in range(1, n + 1)])
````

**Step 3:** Write a function that takes in a numeric grade (0 to 100) and returns the letter grade (no pluses or minuses). In a traditional rubric, that would be as follows:

* 90 or above = A
* 80 - 89 = B
* 70 - 79 = C
* 60 - 69 = D
* 59 or below = F

```
letter_grade(85) = 'B'
```

````python
def letter_grade(grade):
    if grade >= 90:
        return 'A'
    elif grade >= 80:
        return 'B'
    elif grade >= 70:
        return 'C'
    elif grade >= 60:
        return 'D'
    else:
        return 'F'
````

````python
letter_grade(85)
````

**Step 4:** Below is the cipher from Trial 1. Write a function that encodes the given strings as input. Make sure to convert the input string to capslock (coded message output will always be in caps.)

```
encode('python') = 'SBWKRQ'
```

````python
cipher = {
    'A': 'D', 'B': 'E', 'C': 'F', 'D': 'G', 'E': 'H',
    'F': 'I', 'G': 'J', 'H': 'K', 'I': 'L', 'J': 'M',
    'K': 'N', 'L': 'O', 'M': 'P', 'N': 'Q', 'O': 'R',
    'P': 'S', 'Q': 'T', 'R': 'U', 'S': 'V', 'T': 'W',
    'U': 'X', 'V': 'Y', 'W': 'Z', 'X': 'A', 'Y': 'B',
    'Z': 'C'
}
````

````python
def encode(text):
    return ''.join([cipher.get(char, char) for char in text.upper()])
````

````python
encode('python')
````

**Step 5a:** Write a function that takes a string as input and outputs the number of vowels in that string. Y is not a vowel.

```
vowel_count('avocado') = 4
```

````python
def vowel_count(text):
    count = 0
    for char in text.lower():
        if char in 'aeiou':
            count += 1
    return count
````

````python
vowel_count('avocado')
````

**Step 5b:** Write a function that takes a string as input (which may be multiple words long), and reduces each word to its vowel count. So yes, you should use your answer to 5a to solve step b. Don't rewrite code from 5a, simply use the `vowel_count()` function inside this new one.

```
vowel_count_encoder('journey before destination') = '3 3 5'
```

````python
def vowel_count_encoder(text):
    words = text.split()
    return ' '.join([str(vowel_count(w)) for w in words])
````

````python
vowel_count_encoder('journey before destination')
````
